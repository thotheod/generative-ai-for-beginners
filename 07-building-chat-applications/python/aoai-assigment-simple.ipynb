```python
import os
from openai import AzureOpenAI

openai.api_type = "azure"
openai.api_version = os.getenv("AZURE_OPENAI_API_VERSION","").strip()

API_KEY = os.getenv("AZURE_OPENAI_GPT4o_API_KEY","").strip()
assert API_KEY, "ERROR: Azure OpenAI Key is missing"
openai.api_key = API_KEY

RESOURCE_ENDPOINT = os.getenv("AZURE_OPENAI_GPT4o_ENDPOINT","").strip()
assert RESOURCE_ENDPOINT, "ERROR: Azure OpenAI Endpoint is missing"
# assert "openai.azure.com" in RESOURCE_ENDPOINT.lower(), "ERROR: Azure OpenAI Endpoint should be in the form: \n\n\t<your unique endpoint identifier>.openai.azure.com"
openai.api_base = RESOURCE_ENDPOINT
deployment=os.environ['AZURE_OPENAI_GPT4o_DEPLOYMENT']
endpoint=os.environ["AZURE_OPENAI_GPT4o_API_BASE"]
API_KEY = os.environ['AZURE_OPENAI_GPT4o_API_KEY']
assert deployment, "ERROR: Azure OpenAI Deployment name is missing"
```

```python
# # Create your first prompt
# text_prompt = " My foot hurts, what can be wrong?"

# Let's verify the environment variables are loaded
print(f"Endpoint: {endpoint}")
print(f"API Key set?: {'yes' if API_KEY else 'no'}")
print(f"Deployment: {deployment}")


# response = client.chat.completions.create(model=deployment, messages=messages, temperature=0.99, max_tokens=100) 
# response = openai.ChatCompletion.create(
#   engine=deployment,
#   messages = [
#       {"role":"system", "content":"I'm a doctor, specialist on surgery"},
#       {"role":"user","content":text_prompt},])
client = AzureOpenAI(
    azure_endpoint = endpoint,
    api_key = API_KEY,
    api_version = os.environ.get('AZURE_OPENAI_API_VERSION', '2024-02-01')
)

# client = AzureOpenAI(
#     azure_endpoint = os.environ["AZURE_OPENAI_GPT4o_API_BASE"],   # use azure_endpoint
#     api_key = os.environ['AZURE_OPENAI_GPT4o_API_KEY'],
#     api_version = os.environ.get('AZURE_OPENAI_API_VERSION', '2024-02-01')

# add your completion code
prompt = "Complete the following: Once upon a time there was a"
messages = [{"role": "user", "content": prompt}]  

# make completion
completion = client.chat.completions.create(model=deployment, messages=messages, temperature=0.99, max_tokens=100) #the lower the temperature, the more deterministic the output

print(completion.choices[0].message.content)
# response['choices'][0]['message']['content']
```
